#### Method overriding and method overloading are key concepts in object-oriented programming, particularly in languages like Python. They both relate to how methods with the same name are handled, but they serve different purposes and are implemented differently

### Method Overloading
- Method overloading means defining multiple methods with the same name but different parameters. Python does not support method overloading like other programming languages (such as Java and C++), but we can achieve a similar effect using default arguments or the *args and **kwargs parameters.

```python
class Demo:
    def add(self, a, b):
        return a + b

    def add(self, a, b, c):
        return a + b + c

d = Demo()
print(d.add(2, 3))      # This will raise an error
print(d.add(1, 2, 3))   # This works correctly
```

```python
# achieving this using *args
class Demo:
    def add(self, *args):
        c=0
        for i in args:
            c+=i
        return c

d = Demo()
print(d.add(2, 3))      
print(d.add(1, 2, 3)) 
```

```python

class Rectangle:
    def __init__(self, *args):
        if len(args) == 1:  # Assume it's a square
            self.width = self.height = args[0]
        elif len(args) == 2:  # Width and height
            self.width, self.height = args
        else:
            raise ValueError("Invalid arguments")

    def area(self):
        return self.width * self.height

r1 = Rectangle(5)      # Square
print(r1.area())       # 25

r2 = Rectangle(3, 4)   # Rectangle
print(r2.area())       # 12
```

### Method Overriding
- Method overriding occurs when a subclass (child class) provides a specific implementation of a method that is already defined in its superclass (parent class).
- It is possible only in inheritance...With single class, over riding is not possible

```python
class Animal:
    def sound(self):
        print("Some generic animal sound")
class Dog(Animal):
    def sound(self):
        print("Bark")
class Cat(Animal):
    def sound(self):
        print("Meow")
a = Animal()
d = Dog()
c = Cat()
a.sound()   # Output: Some generic animal sound
d.sound()   # Output: Bark
c.sound()   # Output: Meow
```

```python
class Dog(Animal):
    def sound(self):
        super().sound()  # Call parent version
        print("Bark")    # Add more behavior
d = Dog()
d.sound()
```

```python
#Example 2nd
class Father:
    def sleep(self):
        print("sleeps from 10:00 PM to 5:00 AM")

    def eat(self):
        print("eating")

class Son(Father):
    def sleep(self):
        print("sleeps from 2:00 AM to 10:00 AM")

obj = Son()
obj.sleep()
obj.eat()
```
